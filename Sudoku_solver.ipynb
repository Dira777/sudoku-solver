```python
import random
N = 9  # 9x9 Grid
def print_grid(grid):
    for i in range(N):
        if i % 3 == 0 and i != 0:
            print("-" * 21)
        for j in range(N):
            if j % 3 == 0 and j != 0:
                print("|", end=" ")
            print(grid[i][j] if grid[i][j] != 0 else ".", end=" ")
        print()
    print()
# Function to check if the number is valid for that position
def is_safe(grid, row, col, num):
    for x in range(N):
        if grid[row][x] == num or grid[x][col] == num:
            return False
    start_row, start_col = row - row % 3, col - col % 3
    for i in range(3):
        for j in range(3):
            if grid[i + start_row][j + start_col] == num:
                return False
    return True
# Function to check if the puzzle is completely and correctly solved
def is_solved(grid):
    for i in range(N):
        for j in range(N):
            if grid[i][j] == 0:
                return False
            num = grid[i][j]
            grid[i][j] = 0  
            if not is_safe(grid, i, j, num):
                grid[i][j] = num
                return False
            grid[i][j] = num
    return True
# Function to check if the puzzle has at least one valid solution 
def is_solvable(grid):
    empty = []
    for i in range(N):
        for j in range(N):
            if grid[i][j] == 0:
                empty.append((i, j))
    
    def solve(pos):
        if pos == len(empty):
            return True
        row, col = empty[pos]
        for num in range(1, 10):
            if is_safe(grid, row, col, num):
                grid[row][col] = num
                if solve(pos + 1):
                    grid[row][col] = 0  # Reset after check
                    return True
                grid[row][col] = 0
        return False
    
    return solve(0)
# Function to generate a valid random puzzle
def generate_random_puzzle(filled_cells=20):
    while True:
        grid = [[0 for _ in range(N)] for _ in range(N)]
        count = 0
        attempts = 0
        while count < filled_cells and attempts < 200:
            i, j = random.randint(0, 8), random.randint(0, 8)
            num = random.randint(1, 9)
            if grid[i][j] == 0 and is_safe(grid, i, j, num):
                grid[i][j] = num
                count += 1
            attempts += 1
        if is_solvable(grid):  # Only return if puzzle is solvable
            return grid
# Main game function
def play_sudoku(puzzle):
    print("Welcome to Sudoku!")
    print("Fill in the empty cells (represented by '.')")
    print("Type 'q' at any time to quit.\n")

    while True:
        print_grid(puzzle)
        try:
            row_input = input("Enter row (1-9) or 'q' to quit: ")
            if row_input.lower() == 'q':
                print("Thanks for playing! See you next time.")
                break

            col_input = input("Enter column (1-9) or 'q' to quit: ")
            if col_input.lower() == 'q':
                print("Thanks for playing! See you next time.")
                break

            num_input = input("Enter number (1-9) or 'q' to quit: ")
            if num_input.lower() == 'q':
                print("Thanks for playing! See you next time")
                break

            row = int(row_input) - 1
            col = int(col_input) - 1
            num = int(num_input)

            # Checks whether the input is within range
            if not (0 <= row < N and 0 <= col < N and 1 <= num <= N):
                print("Invalid input! Please enter values within range.\n")
                continue

            # Checks if the cell is already filled
            if puzzle[row][col] != 0:
                print("Cell already filled! Choose an empty cell.\n")
                continue

            # Checks whether the move follows Sudoku rules
            if is_safe(puzzle, row, col, num):
                puzzle[row][col] = num
                print("Move accepted!\n")
            else:
                print("Invalid move. Number conflicts Sudoku rules.\n")
                continue

            # Checks if the puzzle is fully and correctly solved
            if is_solved(puzzle):
                print_grid(puzzle)
                print("Great! You solved the Sudoku!")
                break

        # Handle non-integer inputs
        except ValueError:
            print("Please enter valid integers!\n")

puzzle = generate_random_puzzle(filled_cells=20)
play_sudoku(puzzle)
```
